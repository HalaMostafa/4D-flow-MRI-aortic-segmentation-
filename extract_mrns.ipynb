```python
import os
import pandas as pd
from pathlib import Path
```

```python
BASE_DIR = os.path.dirname(os.path.abspath(os.getcwd()))
RAW_DATA_DIR = os.path.join(BASE_DIR, 'workspace', '4Dflows')

subfolders = [
    name for name in os.listdir(RAW_DATA_DIR)
    if os.path.isdir(os.path.join(RAW_DATA_DIR, name))
]

df = pd.DataFrame(subfolders, columns=['ID'])
df.to_csv('ids.csv', index=False)

print(f"Saved {len(subfolders)} IDs to ids.csv")
```
